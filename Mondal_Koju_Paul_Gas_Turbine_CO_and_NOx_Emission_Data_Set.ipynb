# Load data

```python
import sys
import os
import zipfile
import urllib.request

DOWNLOAD_ROOT = "https://archive.ics.uci.edu/ml/machine-learning-databases/00551/"
TURBINE_PATH =  os.path.join("Data Folder","Dataset")
TURBINE_URL = DOWNLOAD_ROOT + "pp_gas_emission.zip"

def fetch_turbine_data (turbine_url = TURBINE_URL, turbine_path = TURBINE_PATH):
  if not os.path.isdir(turbine_path):
    os.makedirs(turbine_path)
  zip_path = os.path.join(turbine_path, "pp_gas_emission.zip")
  urllib.request.urlretrieve(turbine_url, zip_path)
  with zipfile.ZipFile(zip_path,'r') as myzip:
    dataset_path = os.path.join(turbine_path, "Dataset")
    myzip.extractall('Datasets')
```

# download the file

```python
fetch_turbine_data()
```

# Polynomial Regression

```python
from sklearn.preprocessing import PolynomialFeatures
poly_reg = PolynomialFeatures(degree = 3)
```

```python
from sklearn.model_selection import train_test_split

X1_poly_train, X1_poly_test, y1_poly_train, y1_poly_test = train_test_split(X, y1, test_size = 0.2, random_state = 0)
X2_poly_train, X2_poly_test, y2_poly_train, y2_poly_test = train_test_split(X, y2, test_size = 0.2, random_state = 0)
```

```python
lin_reg_1 = LinearRegression()

X1_polytrain_new = poly_reg.fit_transform(X1_poly_train)
lin_reg_1.fit(X1_polytrain_new, y1_poly_train)

X2_polytrain_new = poly_reg.fit_transform(X2_poly_train)
lin_reg_2 = LinearRegression()
lin_reg_2.fit(X1_polytrain_new, y2_poly_train)

X1_polytest_new = poly_reg.fit_transform(X1_poly_test)
X2_polytest_new = poly_reg.fit_transform(X2_poly_test)
y1_poly_pred = lin_reg_1.predict(X1_polytest_new)
y2_poly_pred = lin_reg_2.predict(X2_polytest_new)
```

```python
plt.scatter(X1_poly_test['AT'],y1_poly_test)
plt.plot(X1_polytest_new[:,1], y1_poly_pred, 'rx')
plt.xlabel('Ambient temperature (AT)')
plt.ylabel('Carbion monoxide (CO)')
plt.legend(["predicted","real"])
```

```python
plt.scatter(X1_poly_test['GTEP'],y1_poly_test)
plt.plot(X1_polytest_new[:,5], y1_poly_pred, 'rx')
plt.xlabel('Gas turbine exhaust pressure (GTEP)')
plt.ylabel('Carbion monoxide(CO)')
plt.legend(["predicted","real"])
```

```python
plt.scatter(X1_poly_test['AT'],y2_poly_test)
plt.plot(X1_polytest_new[:,1], y2_poly_pred, 'rx')
plt.xlabel('Ambient temperature (AT)')
plt.ylabel('Nitrogen oxides (NOx)')
plt.legend(["predicted","real"])
```

```python
plt.scatter(X1_poly_test['GTEP'],y2_poly_test)
plt.plot(X1_polytest_new[:,5], y2_poly_pred, 'rx')
plt.xlabel('Gas turbine exhaust pressure (GTEP')
plt.ylabel('Nitrogen oxides (NOx)')
plt.legend(["predicted","real"])
```

Statistical Analysis
