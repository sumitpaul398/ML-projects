```python
#%% Random data generator
import numpy as np
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt

n = 100
p = 1
alpha = 0.2
n_samples, n_features = n, p
rng = np.random.RandomState(0)
noise = rng.normal(loc=0.0, scale=10, size=n_samples)
X, y = make_regression(n_samples, n_features, random_state=rng)
y = y + noise
```

```python
#%% Gradient descent solver
beta = 0
c = 0

for i in range(1000):
    gd_beta = 1/n * sum([ (y[i] - X[i,0]*beta - c) * (-X[i,0]) for i in range(n)])
    gd_c = 1/n * sum([(y[i] - X[i,0]*beta - c) * (-1) for i in range(n)])
    beta = beta - alpha*gd_beta
    c = c - alpha*gd_c
y_gd = X[:,0]*beta + c
```

```python
#%% Exact solver
X_exact = np.append(X, np.ones((n,1)),axis=1)
beta_exact = np.linalg.inv(np.transpose(X_exact) @ X_exact) @ np.transpose(X_exact) @ y
y_exact = X_exact @ beta_exact
```

```python
plt.scatter(X[:,0],y)
plt.plot([min(X[:,0]),max(X[:,0])], [min(y_gd), max(y_gd)], color = 'red', marker ="x")
plt.plot([min(X[:,0]),max(X[:,0])], [min(y_exact), max(y_exact)], color = 'blue', marker = "^")
plt.show()
#So X[:,0] - > would say list elements in all rows as it just colon : present in first column so the column of entire matrix is printed out. dimension is [no_of_rows * 1]
#Similarly, X[:,1] - > this would list the second column from all rows.

```

```python
# %% Print beta and c

print("Beta from gradient descent is {}".format(beta))
print("Beta from exact solver is {}".format(beta_exact[0]))
print("C from gradient descent is {}".format(c))
print("C from exact solver is {}".format(beta_exact[1]))

# %%
```
